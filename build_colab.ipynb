```python
# Cell 1 - Upload your zip and extract
from google.colab import files
import zipfile, os
uploaded = files.upload()
for fname in uploaded:
    with zipfile.ZipFile(fname, 'r') as z:
        z.extractall('/content')
print('✅ Files extracted!')
os.listdir('/content')
```

```python
%%bash
# Cell 2 - Install dependencies
apt-get install -y python3-pip build-essential git \
    libsdl2-dev libsdl2-image-dev libsdl2-mixer-dev libsdl2-ttf-dev \
    libportmidi-dev libswscale-dev libavformat-dev libavcodec-dev \
    zlib1g-dev openjdk-17-jdk unzip zip -q
pip install buildozer cython==0.29.33 -q
echo '✅ Dependencies installed!'
```

```python
%%bash
# Cell 3 - Build APK (takes 15-25 mins first time)
cd /content
rm -rf .buildozer bin/
buildozer android debug 2>&1 | tail -80
```

```python
# Cell 4 - Download APK
import glob, os
from google.colab import files
apks = glob.glob('/content/bin/*.apk')
if apks:
    print(f'✅ APK ready: {apks[0]}')
    print(f'   Size: {os.path.getsize(apks[0])/1024/1024:.1f} MB')
    files.download(apks[0])
else:
    print('❌ APK not found - check Cell 3 output for errors')
```
